```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter


loader = TextLoader('speech.txt')
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size = 1000, chunk_overlap = 30)
docs = text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")

db = FAISS.from_documents(docs, embeddings)
db
```

```python
## query
query = "How does the speaker describe the desired outcome of the war"

docs = db.similarity_search_with_score(query)
docs
```

```python
docs[0].page_content
```

```python
## retrieval
retriver = db.as_retriever()
docs = retriver.invoke(query)
docs
```

```python
embeddings_query = embeddings.embed_query(query)
docs = db.similarity_search_by_vector(embeddings_query)
docs
```

```python
## saving & loading

db.save_local("faiss_index")
```

```python
new_df = FAISS.load_local("faiss_index",embeddings, allow_dangerous_deserialization=True)
docs = new_df.similarity_search(query)
docs
```
